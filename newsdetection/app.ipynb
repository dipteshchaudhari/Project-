🔎 What is Training and Testing in Machine Learning?

Training Data

This is the part of the dataset we use to teach the model.

The model looks at the input (text) and the output (class) to learn patterns.

Example:

Input: "Trump wins election" → Output: 1 (True news)

Input: "Aliens attacked the White House" → Output: 0 (Fake news)

By seeing many such examples, the model learns what fake and true news look like.

Testing Data

This is the part of the dataset that the model has never seen before.

We use it to check how well the model performs on new, unseen news.

Example: After training, we give it a new text "Government launches new scheme" → The model predicts if it’s Fake or True.

If the model gets it right most of the time, we know it has learned wel

in below code , We are spitting four parts :

******xtrain (Training Features)*******

These are the input news texts (from "text") used for training the model.
Example: "Government announces new policy" → model will read this.

*******ytrain (Training Labels)*******

These are the correct answers (from "class") for the training inputs.

Example: 1 (True news), 0 (Fake news).

The model uses these labels to learn the relationship between text and class.

*******xtest (Testing Features)*******

These are the input news texts reserved for testing (25% of the data).

The model has never seen these before.

********ytest (Testing Labels)********

These are the correct answers for the testing inputs.

We don’t show these to the model during training.

After training, we compare the model’s predictions with ytest to calculate accuracy, precision, recall, etc.



"xtrain and ytrain are like the practice material for the model: it learns patterns from them.

xtest and ytest are like the final exam: we test the model on new data (xtest) and check if its answers match the real answers (ytest)."

Step 1.2 : 

The Upgrade: We are replacing our old method (TF-IDF), which is just a word counter, with a powerful "brain" called BERT.

The Job: BERT will read each news article and create a special "meaning barcode" for it.


The Result: This "barcode" is a list of numbers that captures the article's real context and meaning. Using these smart barcodes will make our fake news detector much more accurate


"meaning barcode" in column -> bert_embedding


Inside this column, each news article is now represented by a "meaning barcode"—a list of 768 numbers (a vector).

This barcode is the final, tangible result. You have successfully converted raw, messy text into a clean, intelligent, numerical format that a machine can understand.


#Line 1: tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
What it does: This line downloads the "Instruction Manual" (the Tokenizer).

The job of the tokenizer is to take a normal sentence and translate it into a special numerical code that the BERT "brain" can understand. It breaks words into smaller pieces and converts them into numbers.

The command .from_pretrained means we are downloading a pre-made manual that perfectly matches the brain we are about to get.


#Line 2: model = TFBertModel.from_pretrained('bert-base-uncased', use_safetensors=False)
What it does: This line downloads the actual "Brain" (the TFBertModel).

The command .from_pretrained is the most important part. It tells the program to download a brain that has already been trained for thousands of hours on a massive amount of text from the internet. This is why it's so smart from the start and can give us a "superior understanding of language context".


'bert-base-uncased' is the specific version we're downloading: base is the standard, medium size, and uncased means it treats "The" and "the" as the same word.

use_safetensors=False is the special instruction we added to fix the previous error. It tells the program to download the model in the older, classic file format, which we know works correctly on your system.

```python
import pandas as pd
import tensorflow as tf
from transformers import BertTokenizer, TFBertModel
import numpy as np


print("Loading data...")
df_fake = pd.read_csv("Fake.csv")
df_true = pd.read_csv("True.csv")
df_fake["class"] = 0
df_true["class"] = 1
data = data.sample(frac=1, random_state=42).reset_index(drop=True)
print("Data loaded.")

# 1. Load the pre-trained BERT model and its tokenizer.
print("Loading BERT tokenizer and model...")
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = TFBertModel.from_pretrained('bert-base-uncased', use_safetensors=False)
print("BERT model loaded.")

# 2. Create a function to get the embedding for a single piece of text.
def get_bert_embedding(text):
    text = str(text)
    inputs = tokenizer(text, return_tensors='tf', truncation=True, padding=True, max_length=512)
    outputs = model(inputs)
    embedding = tf.reduce_mean(outputs.last_hidden_state, axis=1)
    return embedding.numpy().flatten()

# 3. CRITICAL: Create a very small sample to test the process.
#    Let's use just 10 articles first.
print("\nCreating a tiny sample of 10 articles to test the embedding process...")
sample_data = data.head(10).copy()

# 4. Apply the function to the sample data.
#    BE PATIENT: Even for 10 articles, this may take several minutes on a CPU.
print("Generating BERT embeddings for the sample... This will be slow.")
sample_data['bert_embedding'] = sample_data['text'].apply(get_bert_embedding)
print("Sample embeddings generated successfully!")

# 5. Display the result to see the new column.
print("\nDataFrame with the new 'bert_embedding' feature:")
print(sample_data[['text', 'bert_embedding', 'class']].head())
```

 BERT : Bidirectional Encoder Representations from Transformers

BERT is a revolutionary, pre-trained "brain" for understanding human language, developed by Google.
BERT's main superpower is its ability to understand context.

Older models read a sentence from left-to-right or right-to-left. BERT is bidirectional, meaning it reads the entire sentence at once, looking at all the words simultaneously.

basically it is a superpower

How We Are Using It in Our Project:

In our project, we are using BERT's deep understanding of language to perform one specific task:
We feed it a news article.

It reads the entire article, understanding the context of every word.

It then outputs a "meaning barcode" (a contextual embedding) for that article.

```python
# logistic regression

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
import numpy as np


print("Preparing data for the model...")
# 1. Define our features (X) and our labels (y).
#    CORRECTED: Using 'sample_data' which you created in the previous cell.
X = sample_data['bert_embedding']
y = sample_data['class']

# 2. Convert the 'bert_embedding' column into a format scikit-learn can use.
X_stacked = np.vstack(X.values)

# 3. Split the data into a training set and a testing set.
X_train, X_test, y_train, y_test = train_test_split(X_stacked, y, test_size=0.25, random_state=42, stratify=y)
print(f"Data split into {len(X_train)} training samples and {len(X_test)} testing samples.")

# 4. Initialize and train the Logistic Regression model.
print("\nTraining the Logistic Regression model...")
lr_model = LogisticRegression(max_iter=1000) 
lr_model.fit(X_train, y_train)
print("Model training complete.")

# 5. Evaluate the model on the test data.
print("\nEvaluating the model on the test set...")
predictions = lr_model.predict(X_test)

# 6. Print the classification report to see the results with detailed metrics.
print("\nClassification Report:")
print(classification_report(y_test, predictions))
```

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from transformers import BertTokenizer, TFBertModel
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

# ==============================================================================
# PART 1: LOAD AND PREPARE DATA
# ==============================================================================
print("Step 1: Loading the 'combined_social_data.csv' dataset...")
# Load the correct dataset
data = pd.read_csv("combined_social_data.csv")

# Ensure the data is shuffled (it should al ready be, but this is good practice)
data = data.sample(frac=1, random_state=42).reset_index(drop=True)
print("Data loaded successfully.")

# ==============================================================================
# PART 2: GENERATE BERT EMBEDDINGS (THE SLOW PART)
# ==============================================================================
print("\nStep 2: Generating BERT embeddings for a 2,000-article sample...")
print("This part will take a significant amount of time. Please be patient.")

# Load BERT model and tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = TFBertModel.from_pretrained('bert-base-uncased', use_safetensors=False)

# Define the embedding function
def get_bert_embedding(text):
    text = str(text)
    inputs = tokenizer(text, return_tensors='tf', truncation=True, padding=True, max_length=512)
    outputs = model(inputs)
    embedding = tf.reduce_mean(outputs.last_hidden_state, axis=1)
    return embedding.numpy().flatten()

# Create the 2,000-article sample from our loaded data
sample_data = data.head(2000).copy()

# Apply the function to generate embeddings
sample_data['bert_embedding'] = sample_data['text'].apply(get_bert_embedding)
print("BERT embeddings generated successfully!")

# ==============================================================================
# PART 3: PREPARE VARIABLES FOR THE MODEL
# ==============================================================================
print("\nStep 3: Preparing the final variables for model training...")
X = sample_data['bert_embedding']
y = sample_data['class']    
X_stacked = np.vstack(X.values)
print("Variables 'X_stacked' and 'y' are now ready.")

# ==============================================================================
# PART 4: HYPERPARAMETER TUNING
# ==============================================================================
print("\nStep 4: Starting hyperparameter tuning...")

# Define the model and parameter grid
model_to_tune = LogisticRegression(max_iter=2000)
param_grid = {
    'C': [0.1, 1, 10, 100],
    'solver': ['liblinear', 'saga']
}

# Set up and run the GridSearchCV
grid_search = GridSearchCV(estimator=model_to_tune, param_grid=param_grid, cv=5, scoring='accuracy', verbose=1)
grid_search.fit(X_stacked, y)
print("Search complete.")

# Print the final results
print("\n-------------------------------------------")
print("Tuning Results:")
print("The best settings found by the search are:")
print(grid_search.best_params_)

print("\nThe best accuracy score achieved with these settings was:")
print(f"{grid_search.best_score_:.4f}")
print("-------------------------------------------")
```
